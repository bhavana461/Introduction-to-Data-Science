```python
# Question 7 You are given a customer review: review = "The product quality is Excellent but delivery was Late" Write Python code to: 1. Convert to lowercase 2. Count number of words 3. Check if the word "excellent" exists 4. Replace "Late" with "delayed"
review = "The product quality is Excellent but delivery was Late"

# 1. Convert to lowercase
lower_review = review.lower()
print("Lowercase:", lower_review)

# 2. Count words
word_count = len(review.split())
print("Word Count:", word_count)

# 3. Check if excellent exists
print("excellent" in lower_review)

# 4. Replace Late with delayed
updated_review = review.replace("Late", "delayed")
print(updated_review)
```

```python
# Question 8: Given sales data: sales = [12000, 15000, 11000, 18000, 17000] 1. Calculate total revenue. 2. Find average revenue. 3. Identify highest sales month. 4. Append new month sale = 20000. 5. Sort the list in descending order.
sales = [12000, 15000, 11000, 18000, 17000]

# Total Revenue
total_revenue = sum(sales)
print("Total Revenue:", total_revenue)

# Average Revenue
average_revenue = total_revenue / len(sales)
print("Average Revenue:", average_revenue)

# Highest Sales Month
highest_sale = max(sales)
month = sales.index(highest_sale) + 1
print("Highest Sales Month:", month)

# Append New Sale
sales.append(20000)
print("Updated List:", sales)

# Sort Descending
sales.sort(reverse=True)
print("Descending Order:", sales)
```

```python
#Question 9:You are building a simple sales analysis system. Given: products = ["Laptop", "Phone", "Tablet"] prices = [50000, 20000, 15000] quantities = [5, 10, 7] Tasks: 1. Compute total revenue for each product. 2. Identify which product generated maximum revenue. 3. Store revenues in a new list. 4. Print a formatted summary like:
products = ["Laptop", "Phone", "Tablet"]
prices = [50000, 20000, 15000]
quantities = [5, 10, 7]

revenues = []

for i in range(len(products)):
    revenue = prices[i] * quantities[i]
    revenues.append(revenue)

for i in range(len(products)):
    print(products[i], "Revenue = ₹", revenues[i])

max_revenue = max(revenues)
max_index = revenues.index(max_revenue)

print("Maximum Revenue Product:", products[max_index])
```

 Question 4: An e-commerce company wants an AI system that:
● Monitors inventory
● Predicts demand
● Automatically places supplier orders
● Negotiates price via email
Explain:
1. Why this is an example of Agentic AI.
2. What risks might arise?
3. What human oversight would still be required?

Answer:

This system is an example of Agentic AI because it not only analyzes information but also takes actions on its own. It monitors inventory levels, predicts future demand, places supplier orders automatically, and negotiates prices through email. Such autonomy allows the system to achieve business goals with minimal human involvement.

However, several risks may arise. Incorrect demand predictions can lead to excess inventory or stock shortages. The system may negotiate unfavorable prices or make purchasing decisions that increase costs. Security and reliability issues may also affect business operations.

Human oversight is still necessary to review important decisions, approve large purchases, monitor supplier interactions, and ensure that the AI system behaves according to company policies and business objectives.

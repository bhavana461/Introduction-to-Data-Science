Question 1:Explain the complete ML lifecycle. How is it different from traditional Software
Development Lifecycle (SDLC)?

Answer : -  

The Machine Learning (ML) lifecycle is a systematic process used to develop, deploy, and maintain machine learning models. It begins with understanding the business problem and defining clear objectives. After this, relevant data is collected from various sources and prepared through cleaning, transformation, and feature engineering. The prepared data is then used to train machine learning models. Different models are evaluated, and the best-performing model is selected. Once validated, the model is deployed into production where it can make predictions on new data. Continuous monitoring is required to ensure that the model maintains its performance over time. If performance decreases, the model must be retrained using updated data.

Traditional SDLC focuses on developing software based on predefined rules written by developers. In contrast, ML systems learn patterns from data rather than relying solely on manually written rules. Therefore, ML projects require additional stages such as data collection, model training, evaluation, monitoring, and retraining, making them more data-centric than traditional software projects.



Question 2: Why is translating business needs into a data science problem the most critical
step?
Give one example where poor problem framing leads to wasted resources.

Answer :

Translating business needs into a data science problem is the most important step because it ensures that the solution developed by the data science team directly addresses the organization's goals. If the problem is not properly defined, even the most advanced models may fail to provide useful results. A clear problem statement helps identify the right data, select appropriate techniques, and establish meaningful evaluation metrics.

For example, suppose a company wants to increase customer retention. If the data science team interprets the problem as predicting customer purchases instead of identifying customers likely to leave, significant time and resources may be spent developing a model that does not solve the actual business challenge. Poor problem framing can therefore lead to wasted effort, incorrect decisions, and financial losses.

Question 4: Define Structured Data, Unstructured Data, and Multimodal Data. Give examples for each.

Answer:

Structured data refers to data that is organized in a predefined format, typically stored in rows and columns within databases or spreadsheets. Because of its organized structure, it is easy to search, analyze, and process. Examples include customer records, sales transactions, and employee databases.

Unstructured data refers to information that does not follow a predefined format or schema. Such data is often complex and requires specialized techniques for analysis. Examples include emails, social media posts, images, videos, and audio recordings.

Multimodal data combines multiple forms of data such as text, images, audio, and video within a single system. It enables AI models to process information from different sources simultaneously. Examples include self-driving car systems that use camera images, sensor data, and GPS information, or AI assistants that process both speech and text.

Question 5: Dataset contains 20% missing values, duplicate rows, and imbalanced class distribution. Explain step-by-step how you would clean and prepare this dataset.

Answer:

The first step is to inspect the dataset and identify the extent of missing values, duplicate records, and class imbalance. Missing values should be handled by either removing records with excessive missing information or replacing missing values using suitable methods such as mean, median, mode, or advanced imputation techniques.

After handling missing values, duplicate rows should be identified and removed to prevent redundant information from influencing model training. Once duplicates are removed, the dataset should be checked for inconsistencies, incorrect data types, and outliers.

Next, the class distribution should be analyzed. If one class significantly outnumbers another, balancing techniques such as oversampling, undersampling, or synthetic data generation methods like SMOTE can be applied. Finally, the cleaned dataset should be normalized or standardized if required and then split into training, validation, and testing sets before model development.


Question 6: Explain Tokenization, Embeddings, and Context Window. Why are embeddings critical for GenAI systems?

Answer:

 Tokenization is the process of breaking text into smaller units called tokens. These tokens may be words, subwords, or characters that can be processed by machine learning models.

Embeddings are numerical vector representations of words, phrases, or documents. They capture the semantic meaning and relationships between different pieces of text, allowing AI models to understand context and similarity.

A context window refers to the amount of information a language model can consider at one time while processing input. A larger context window enables the model to remember and use more information from the conversation or document.

Embeddings are critical for Generative AI systems because computers cannot directly understand human language. Embeddings convert text into meaningful numerical representations that preserve semantic relationships. This allows AI systems to perform tasks such as information retrieval, recommendation, semantic search, and natural language understanding effectively.


Question 7: Explain Train/Validation/Test Split, Cross-Validation, and Data Leakage. Why is data leakage dangerous?

Answer:

The train, validation, and test split is a common method used to evaluate machine learning models. The training dataset is used to learn patterns from data. The validation dataset is used to tune model parameters and select the best model. The test dataset is used only once to evaluate the final model's performance on unseen data.

Cross-validation is a technique in which the dataset is divided into multiple subsets. The model is trained and tested multiple times using different combinations of these subsets. This provides a more reliable estimate of model performance.

Data leakage occurs when information from outside the training process is unintentionally used during model development. As a result, the model gains access to information that would not be available in real-world situations.

Data leakage is dangerous because it produces overly optimistic evaluation results. A model may appear highly accurate during testing but perform poorly when deployed in production, leading to unreliable predictions and business losses.


Question 8: Explain Accuracy, Precision, Recall, F1-Score, and ROC-AUC. Why is accuracy misleading in fraud detection?

Answer:

Accuracy measures the proportion of correct predictions made by a model out of all predictions. It provides an overall measure of performance but may not be suitable for imbalanced datasets.

Precision measures how many of the instances predicted as positive are actually positive. It is useful when false positives are costly.

Recall measures how many actual positive cases are correctly identified by the model. It is important when missing positive cases can have serious consequences.

F1-Score combines precision and recall into a single metric by calculating their harmonic mean. It provides a balanced evaluation when both false positives and false negatives are important.

ROC-AUC measures the ability of a model to distinguish between classes across different classification thresholds. A higher ROC-AUC value indicates better discrimination capability.

Accuracy can be misleading in fraud detection because fraudulent transactions typically represent a very small portion of all transactions. A model that predicts every transaction as non-fraudulent may still achieve very high accuracy while completely failing to detect actual fraud cases.


Question 9:
After 6 months:
● Model accuracy drops.
● Customer behavior changes.
Explain:
1. Data drift
2. Concept drift
3. Monitoring techniques
4. Retraining strategy

Answer:

Data drift occurs when the statistical properties of incoming data change over time compared to the data used during training. As customer behavior evolves, the model may encounter patterns that differ from those it originally learned.

Concept drift occurs when the relationship between input variables and the target variable changes. In such cases, even if the data appears similar, the underlying behavior being predicted has changed.

Monitoring techniques involve continuously tracking model performance metrics such as accuracy, precision, recall, and prediction distributions. Organizations may also monitor input data characteristics and compare them with historical training data to identify drift.

When significant drift is detected, a retraining strategy should be implemented. This involves collecting recent data, updating the training dataset, retraining the model, evaluating its performance, and redeploying the improved version. Regular retraining helps maintain model accuracy and reliability in changing environments.



Question 10: An organization wants to integrate:
● ML fraud detection
● GenAI automated email responses
Design a high-level system architecture explaining:
1. Data flow
2. Model interaction
3. Human-in-the-loop
4. Monitoring checkpoints

Answer:

In this system, customer transactions and customer communications serve as the primary data sources. Transaction data is sent to the fraud detection machine learning model, which analyzes transaction patterns and assigns a fraud risk score. If a transaction is considered suspicious, it is flagged for review.

Customer emails and inquiries are processed by a Generative AI model that automatically generates context-aware responses. The GenAI system can access relevant customer information and organizational knowledge to provide accurate replies.

Human involvement remains important in the system. Fraud alerts with high risk scores should be reviewed by fraud analysts before taking critical actions. Similarly, sensitive or complex email responses generated by the AI should be reviewed by human agents before being sent to customers.

Monitoring checkpoints should be established throughout the system. The fraud detection model should be monitored for prediction accuracy, false positives, and false negatives. The GenAI model should be monitored for response quality, relevance, safety, and compliance with organizational policies. Continuous monitoring ensures reliable performance and enables timely updates when required.

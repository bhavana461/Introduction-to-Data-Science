Question 1: What is a REST API? Explain GET, POST, PUT, DELETE. Why is FastAPI preferred over Flask for modern APIs?

Answer:

A REST API (Representational State Transfer Application Programming Interface) is a set of rules that allows different applications to communicate with each other over the internet using HTTP requests. REST APIs enable clients and servers to exchange data efficiently and are widely used in web and mobile applications.

GET is used to retrieve data from the server without modifying it. POST is used to send new data to the server and create a new resource. PUT is used to update an existing resource completely. DELETE is used to remove a resource from the server.

FastAPI is preferred over Flask for modern APIs because it provides automatic data validation, built-in API documentation, high performance, asynchronous programming support, and strong type checking. These features make API development faster, more reliable, and easier to maintain.

```python
#Question 2: Create a FastAPI application with:  Root endpoint (/) , /courses GET endpoint , /courses/{course_id} using path parameter Add: ,Query parameter level (optional)
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Welcome to Course API"}

@app.get("/courses")
def get_courses(level: str = None):
    return {"level": level}

@app.get("/courses/{course_id}")
def get_course(course_id: int):
    return {"course_id": course_id}
```

- Difference between Path Parameter and Query Parameter

Answer:- A path parameter is part of the URL path and is used to identify a specific resource. For example, in /courses/101, the value 101 is a path parameter.

A query parameter is added after a question mark in the URL and is used for filtering, sorting, or optional information. For example, /courses?level=beginner uses level as a query parameter.

```python
#Question 3:  Add validation to: ● course_id must be integer > 0 ● level must be one of: beginner, intermediate, advanced ,What happens if validation fails?
from fastapi import FastAPI, Path, Query

app = FastAPI()

@app.get("/courses/{course_id}")
def get_course(
    course_id: int = Path(..., gt=0),
    level: str = Query(None, pattern="^(beginner|intermediate|advanced)$")
):
    return {
        "course_id": course_id,
        "level": level
    }
```

Answer:

The validation rules ensure that course_id is greater than zero and level can only be beginner, intermediate, or advanced.

If validation fails, FastAPI automatically returns a 422 Unprocessable Entity error along with details explaining which validation rule was violated. This helps developers identify and correct invalid requests quickly.

Question 5: Authentication vs Authorization, JWT Authentication, and Stateless Authentication

Answer:

Authentication is the process of verifying the identity of a user. It answers the question, "Who are you?" Authorization determines what resources or actions a user is allowed to access after authentication. It answers the question, "What can you do?"

JWT (JSON Web Token) authentication works by generating a signed token after a user successfully logs in. The client stores this token and sends it with future requests. The server verifies the token and grants access if it is valid.

Stateless authentication is scalable because the server does not need to store session information. All required user information is stored within the token itself, reducing server memory usage and making it easier to scale applications across multiple servers.

Question 8: Difference between Path Parameter, Query Parameter, and Request Body. When Should Each Be Used?

Answer:

A path parameter is used to identify a specific resource and is included directly in the URL path. For example, /courses/10.

A query parameter is used for filtering, sorting, searching, or optional values and appears after a question mark in the URL. For example, /courses?level=advanced.

A request body is used to send structured data to the server, usually when creating or updating resources. It commonly contains JSON data.

Path parameters should be used when identifying a specific resource, query parameters when filtering or customizing requests, and request bodies when sending large amounts of structured information to the server.

```python
#Question 10: Use Pydantic BaseSettings to Store Configuration. Why Should Secrets Never Be Hardcoded?
from pydantic_settings import BaseSettings

class Settings(BaseSettings):
    database_url: str = "postgresql://localhost:5432/course_db"
    secret_key: str = "my_secret_key"
    debug: bool = True

settings = Settings()

print(settings.database_url)
print(settings.secret_key)
print(settings.debug)
```

Answer

BaseSettings allows application configuration values to be loaded from environment variables instead of being directly written into the source code. This helps separate configuration from application logic.

Secrets such as database passwords, API keys, and secret keys should never be hardcoded because they may become exposed if the source code is shared publicly. Hardcoding secrets creates serious security risks and makes managing different deployment environments more difficult. Storing secrets securely in environment variables or secret management systems improves security and maintainability.
